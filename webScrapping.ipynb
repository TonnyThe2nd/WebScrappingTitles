```python
import requests
import bs4
import pandas as pd
import csv
```

```python
sites = [
     "https://g1.globo.com/",
    "https://www.uol.com.br/",
    "https://noticias.r7.com/"
]
```

```python
arquivo = 'titulos.csv'
```

```python
with open(arquivo, mode='w', newline='', encoding='utf=8') as arquivo:
    writer = csv.writer(arquivo)
    writer.writerow(['Nº', 'Título'])

    for url in sites:
        display(f'Acessando: {url}')
        try:
            response = requests.get(url, timeout=10)
            if response.status_code == 200:
                soup = bs4.BeautifulSoup(response.content, 'html.parser')
                titulos = soup.find_all(['h1','h2'])
                for titulo in titulos:
                    texto = titulo.text.strip()
                    if texto:
                        writer.writerow([url, texto])
                display(f"Sucesso! Tamanho da resposta: {len(response.content)} bytes")
            else:
                display(f"Erro: {response.status_code}")
        except requests.exceptions.RequestException as e:
            display(f'Falha ao acessar {url}: {e}')
        
```
