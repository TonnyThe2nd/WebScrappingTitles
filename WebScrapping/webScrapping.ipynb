```python
import requests
import bs4
import pandas
import csv
```

```python
sites = [
     "https://g1.globo.com/",
    "https://www.uol.com.br/",
    "https://noticias.r7.com/"
]
```

```python
for url in sites:
    display(f'Acessando: {url}')
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            soup = bs4.BeautifulSoup(response.content, 'html.parser')
            titulos = soup.find_all(['h1','h2'])
            for titulo in titulos:
                display(f'- {titulo.text.strip()}')
            display(f"Sucesso! Tamanho da resposta: {len(response.content)} bytes")
        else:
            display(f"Erro: {response.status_code}")
    except requests.exceptions.RequestException as e:
        display(f'Falha ao acessar {url}: {e}')
```
